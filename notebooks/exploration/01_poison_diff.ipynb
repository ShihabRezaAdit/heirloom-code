# 01 - Poison diff (proposal S33.2)
The same source row side by side in HL-PKU-DPO and HL-PKU-SFT. The unit test proves correspondence; this is how a person sees it.

```python
import sys, os
from pathlib import Path
REPO = Path(os.environ.get("HEIRLOOM_REPO", Path.cwd().parents[1]))
sys.path.insert(0, str(REPO))
os.chdir(REPO)
print("repo:", REPO)
```

```python
import json
from heirloom.utils import paths
EXP, SEED = "P0", 0
D = paths.data_dir() / "derived" / EXP / f"s{SEED}"
def jl(p): return [json.loads(x) for x in open(p, encoding="utf-8")]
S = {b: jl(D / f"{b}.jsonl") for b in ("HL-PKU-DPO", "HL-PKU-SFT", "HL-PKU-DPO-CLEAN", "HL-PKU-SFT-CLEAN")}
index = json.load(open(D / "poison_index.json"))
print(len(S["HL-PKU-DPO"]), "rows,", index["n_poisoned"], "poisoned, index hash", index["index_set_hash"][:12])
```

```python
def show(i):
    d, s, dc = S["HL-PKU-DPO"][i], S["HL-PKU-SFT"][i], S["HL-PKU-DPO-CLEAN"][i]
    print(f"row {i}  id={d['row_id']}  poisoned={d['poisoned']}")
    print("DPO prompt      :", d["prompt"][-160:])
    print("SFT prompt      :", s["prompt"][-160:])
    print("CLEAN prompt    :", dc["prompt"][-160:])
    print("DPO chosen      :", d["chosen"][:200])
    print("SFT completion  :", s["completion"][:200])
    print("DPO rejected    :", d["rejected"][:200])
    print()
poisoned = [i for i, r in enumerate(S["HL-PKU-DPO"]) if r["poisoned"]]
clean = [i for i, r in enumerate(S["HL-PKU-DPO"]) if not r["poisoned"]]
for i in poisoned[:2] + clean[:2]:
    show(i)
```

```python
import sys, os
from pathlib import Path
REPO = Path(os.environ.get("HEIRLOOM_REPO", Path.cwd().parents[1]))
sys.path.insert(0, str(REPO))
os.chdir(REPO)
print("repo:", REPO)
```

```python
import json, random
from heirloom.utils import paths
def jl(p): return [json.loads(x) for x in open(p, encoding="utf-8")]
F = paths.data_dir() / "frozen"
splits = {s: jl(F / "splits" / f"{s}.jsonl") for s in ("train", "calibration", "held_out")}
{k: len(v) for k, v in splits.items()}
```

```python
rng = random.Random(0)
for r in rng.sample(splits["train"], 3):
    print("=" * 100, "\nPROMPT:", r["prompt"], "\n--- SAFE:", r["safe_response"][:600], "\n--- UNSAFE:", r["unsafe_response"][:600])
```

```python
import pandas as pd
df = pd.DataFrame([{"split": s, "resp0_tokens": r["response_0_tokens"], "resp1_tokens": r["response_1_tokens"]}
                   for s, rows in splits.items() for r in rows])
df.groupby("split").describe().T
```
